```python
EC = LabelEncoder()
data = df.copy()
data['cracking in the weld metal (yes/no)'] = EC.fit_transform(data['cracking in the weld metal (yes/no)'])
df
```

```python
y = data['cracking in the weld metal (yes/no)']
x = data.drop('cracking in the weld metal (yes/no)', axis =1)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state=42)
```

```python
c = DecisionTreeClassifier()
c.fit(x_train,y_train)
```

```python
y_pred = c.predict(x_test)
y_pred
```

```python
y_test
```

```python
accuracy_score(y_pred,y_test)
```
